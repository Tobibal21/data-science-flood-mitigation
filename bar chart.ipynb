```python
import matplotlib.pyplot as plt

# Define data
categories = ['A', 'B', 'C']  # Fixed 'c' to 'C' for consistency
values = [10, 15, 25]

# Create bar chart
plt.bar(categories, values, color=['yellow', 'green', 'chocolate'])

# Labels and title
plt.title('Bar Chart')
plt.xlabel('Categories')  # Fixed spelling
plt.ylabel('Values')  # Fixed spelling

# Show plot
plt.show()
```
